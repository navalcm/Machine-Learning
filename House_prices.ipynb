```python
# Remove rows with missing prices
house_prices = house_prices.dropna(subset=["Price (in rupees)"])
```

Create a Binary Target

```python
median_price = house_prices["Price (in rupees)"].median()

house_prices["Target"] = (
    house_prices["Price (in rupees)"] >= median_price
).astype(int)
```

Fill Missing Values (Categorical Columns)

```python
categorical_columns = house_prices.select_dtypes(include='object').columns

categorical_columns
```

```python
for column in categorical_columns:

    house_prices[column].fillna(
        house_prices[column].mode()[0],
        inplace=True
    )
```

Fill Missing Values (Numeric Columns)

```python
numeric_columns = house_prices.select_dtypes(include=['int64','float64']).columns

numeric_columns
```

```python
for column in numeric_columns:

    house_prices[column].fillna(
        house_prices[column].mean(),
        inplace=True
    )
```

Create Binary Target:

Your dataset is for house prices, so we'll convert the price into a binary target.

```python
median_price = house_prices["Price (in rupees)"].median()

median_price
```

```python
house_prices["Target"] = np.where(
    house_prices["Price (in rupees)"] >= median_price,
    1,
    0
)
```
